#### Filling blank rows

#### Date Format

```python
df['Order_Date'] = pd.to_datetime(
    df['Order_Date'],
    dayfirst=True,
    errors='coerce'
)
print(df['Order_Date'].dtype)
```

```python
df['Year'] = df['Order_Date'].dt.year
df['Month'] = df['Order_Date'].dt.month
df['Month_Name'] = df['Order_Date'].dt.month_name()
df['Quarter'] = df['Order_Date'].dt.quarter
df['Day'] = df['Order_Date'].dt.day
df['Day_Name'] = df['Order_Date'].dt.day_name()
```

```python
df[['Order_Date', 'Year', 'Month', 'Month_Name', 
   'Quarter', 'Day', 'Day_Name']].head()
```

```python
print(df.head(10))
```

```python
df['Order_Date'].isna().sum()
```

##### Customer_ID

```python
df['Customer_ID'] = df['Customer_ID'].fillna('Unknown')
print(df['Customer_ID'].head(30).isna())
```

```python
print(df['Customer_ID'].isnull().sum())
```

##### Category

```python
df['Category'] = df['Category'].str.strip().str.title()
df['Category'] = df['Category'].fillna('Unknown')
print(df.iloc[100])
```

```python
print(df['Category'].isnull().sum())
```

```python
print(df.iloc[6])
```

##### Region

```python
df['Region'] = df['Region'].str.strip().str.title()
df['Region'] = df['Region'].fillna('Unknown')
print(df)
```

```python
print(df['Region'].isnull().sum())
```

##### Quantity

```python
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
print(df['Quantity'].dtype)
```

```python
df[df['Quantity'] <= 0][
    ['Order_ID', 'Product', 'Category', 'Revenue', 'Profit', 'Quantity']
].head(20)
```

```python
df.loc[df['Quantity'] <= 0, 'Quantity'] = np.nan
```

```python
print("Missing Quantity:", df['Quantity'].isna().sum())
```

```python
df['Quantity'] = df.groupby('Product')['Quantity'].transform(
    lambda x: x.fillna(x.median())
)
df['Quantity'] = df['Quantity'].round().astype(int)
print(df.head(30))
```

```python
print("Missing Quantity:", df['Quantity'].isna().sum())
print("Zero Quantity:", (df['Quantity'] == 0).sum())
print("Negative Quantity:", (df['Quantity'] < 0).sum())
```

##### Revenue

```python
print(df['Revenue'].dtype)
```

```python
df['Revenue'] = (
    df['Revenue']
    .astype('string')
    .str.strip()
    .str.replace(',', '', regex=False)
    .str.replace(r'\s*INR\s*', '', regex=True)
)
df['Revenue'].head(10)
```

```python
df['Revenue'] = pd.to_numeric(
    df['Revenue'],
    errors='coerce'
).astype(float)
print(df['Revenue'].dtype)
```

```python
print("Missing Revenue:", df['Revenue'].isna().sum())
```

```python
df[df['Revenue'].isna()][
    ['Order_ID', 'Revenue', 'Product', 'Quantity', 'Profit']
].head(30)
```

```python
print("Minimum Revenue:", df['Revenue'].min())
print("Maximum Revenue:", df['Revenue'].max())
```

```python
print("Negative Revenue:", (df['Revenue'] < 0).sum())
print("Zero Revenue:", (df['Revenue'] == 0).sum())
```

```python
df['Revenue'] = df.groupby('Product')['Revenue'].transform(
    lambda x: x.fillna(x.median())
)
df['Revenue'] = df['Revenue'].fillna(df['Revenue'].median())
```

```python
print("Total Rows:", len(df))
print("Missing Revenue:", df['Revenue'].isna().sum())
```

##### Profit

```python
print(df['Profit'].dtype)
```

```python
df['Profit'] = pd.to_numeric(
    df['Profit'],
    errors='coerce'
)
print(df['Revenue'].dtype)
```

```python
print("Negative Profit:", (df['Profit'] < 0).sum())
```

```python
print("Missing Profit:", df['Profit'].isna().sum())
```

```python
df[df['Profit'].isna()][
    ['Order_ID', 'Product', 'Category', 'Region', 'Quantity', 'Revenue', 'Profit']
].head(20)
```

```python
df['Profit'] = df.groupby('Product')['Profit'].transform(
    lambda x: x.fillna(x.median())
)
df['Profit'] = df['Profit'].fillna(df['Profit'].median())
```

```python
print("Missing Profit:", df['Profit'].isna().sum())
print("Total Rows:", len(df))
print("Profit dtype:", df['Profit'].dtype)
```
